```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Part A — Uniform Distribution
# Generate 1,000 random waiting times following a uniform distribution between 3 and 15 minutes.
# Store the generated values in a NumPy array.
# Display the first 10 generated values.
# Calculate and print:
# Mean
# Median
# Standard deviation

NormalValues = np.random.randint(3, 16, 1000)
print(NormalValues[:10])
print(np.mean(NormalValues))
print(np.median(NormalValues))
print(np.std(NormalValues))
```

```python
# Part B — Normal Distribution
# Generate 1,000 waiting times following a normal distribution with:
# Mean = 9 minutes
# Standard deviation = 2 minutes
# Generate the data using NumPy.
# Replace any negative waiting time with 0.
# Calculate and print:
# Mean
# Median
# Standard deviation
# Maximum waiting time

val = np.random.normal(9, 2, 1000)
for x in val:
    if x < 0:
        x = 0
print(np.mean(val))
print(np.median(val))
print(np.std(val))
print(np.max(val))
```

```python
# Part C — Visualization
# Create one figure containing two subplots:
# Subplot 1 → Histogram of the uniform-distribution data
# Subplot 2 → Histogram of the normal-distribution data
# Each plot must have:
# Appropriate title
# X-axis label
# Y-axis label


random_numbers_u = np.random.rand(100000)
random_numbers_n = np.random.randn(100000)

fig, ax = plt.subplots(1, 2)
fig.suptitle('Uniform and Normal Distribution')
ax[0].hist(random_numbers_u)
ax[0].set_title('Uniform Distribution')
ax[0].set_xlabel('Value')
ax[0].set_ylabel('Frequency')

ax[1].hist(random_numbers_n)
ax[1].set_title('Normal Distribution')
ax[1].set_xlabel('Value')
ax[1].set_ylabel('Frequency')
plt.show()
```
