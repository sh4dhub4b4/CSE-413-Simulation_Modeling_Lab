# Module 1: Matrix Operations, Linear Algebra & Student Task
**Source Resources:** `Student_Copy_SIM_Lab_Day_2(matrix_op_graph_plot_subplot).ipynb`, `Lab Assignment 2 Questions.pdf`

### When to Use:
Matrix manipulations form the mathematical backbone of system modeling, linear transformations, discrete state representations, Markov transition matrices, and network simulations.
* **Shape & Size:** To verify matrix dimensions before matrix multiplication ($m \times k$ with $k \times n$).
* **Determinant ($\det(A)$):** Measures the scaling factor of the transformation; non-zero determinant indicates an invertible (non-singular) system.
* **Matrix Rank:** Indicates the maximum number of linearly independent row or column vectors (determines if a system of equations has a unique solution).
* **Eigenvalues & Eigenvectors:** Vital in dynamical systems, state stability analysis, Markov chains, and principal component analysis.
* **Matrix Inversion ($A^{-1}$):** Required to solve systems of linear equations $Ax = b \implies x = A^{-1}b$.

### Class Example & Individualized Student Task:
In Lab Assignment 2, each student forms a matrix using their student ID digits $d_1$ (last digit) and $d_2$ (second-to-last digit), computes its properties, perturbs one element ($+1$), and observes the sensitivity of the determinant, rank, and eigenvalues.


```python
# 1.1 Fundamental Matrix Constructions
I = np.eye(3)        # 3x3 Identity Matrix
Z = np.zeros((3, 3))  # 3x3 Zero Matrix
O = np.ones((2, 3))   # 2x3 Matrix of Ones

print("Identity Matrix:\n", I)
print("Zero Matrix:\n", Z)

# 1.2 Individualized Matrix Analysis (Student ID Task: e.g., ID ending in d2=6, d1=1)
d1 = 1  # Last digit of student ID
d2 = 6  # Second-to-last digit of student ID

A = np.array([
    [d1 + 2, d2, 1],
    [3, d1 + 1, 4],
    [2, 5, d2 + 1]
], dtype=float)

print("\n--- Matrix A Analysis ---")
print("Matrix A:\n", A)
print("Shape of A:", A.shape)

# Determinant
det_A = np.linalg.det(A)
print(f"Determinant of A: {det_A:.4f}")

# Rank
rank_A = np.linalg.matrix_rank(A)
print("Rank of A:", rank_A)

# Eigenvalues and Eigenvectors
eigvals_A, eigvecs_A = np.linalg.eig(A)
print("Eigenvalues of A:", np.round(eigvals_A, 4))

# Matrix Inverse (Valid only if det != 0)
if abs(det_A) > 1e-6:
    inv_A = np.linalg.inv(A)
    print("Inverse of A:\n", np.round(inv_A, 4))
else:
    print("Matrix A is singular and cannot be inverted.")

# 1.3 Value Change Experiment (Matrix B: Add +1 to element [0, 0])
B = A.copy()
B[0, 0] += 1.0  # State explicitly: changed element at index (0,0)

print("\n--- Matrix B (Perturbed Element [0,0] by +1) ---")
print("Matrix B:\n", B)
det_B = np.linalg.det(B)
rank_B = np.linalg.matrix_rank(B)
eigvals_B, _ = np.linalg.eig(B)
print(f"Determinant of B: {det_B:.4f} (Change: {det_B - det_A:+.4f})")
print("Rank of B:", rank_B)
print("Eigenvalues of B:", np.round(eigvals_B, 4))
```

# Module 3: One-Sample $t$-Test
**Source Resources:** `T test.pptx.pdf`, `T_test.ipynb`

### When to Use:
Use a **One-Sample $t$-Test** when:
1. You have a **single sample** of continuous numerical data.
2. You want to test whether the population mean ($\mu$) is significantly different from a **known standard, benchmark, or historical population value** ($\mu_0$).
3. The population standard deviation ($\sigma$) is unknown and the sample size is relatively small ($n < 30$), though it holds for larger samples as well.

### Mathematical Formulation:
$$t = \frac{\bar{x} - \mu_0}{s / \sqrt{n}}$$
* Degrees of Freedom: $df = n - 1$
* Standard Error: $\text{SE} = \frac{s}{\sqrt{n}}$
* Assumptions: Continuous scale, approximately normal distribution, independent random sampling.
* Decision Rule: At significance level $\alpha = 0.05$, reject $H_0$ if $p\text{-value} < \alpha$ or if $|t_{\text{calc}}| > t_{\text{crit}}$.

### Class Examples:
1. **Teacher's Benchmark Problem:** Exam scores of 10 students tested against school average of 70.
2. **Student Work Task:** Measurements `[5.2, 5.4, 5.1, 5.6, 5.3, 5.8]` tested against hypothesized mean of `5.5`.


```python
# 3.1 Class Example: 10 Exam Scores vs Benchmark of 70
# Step 1: Define the sample dataset
exam_scores = np.array([65, 72, 68, 75, 80, 62, 71, 78, 74, 69])
population_mean = 70.0  # H0: mu = 70, H1: mu != 70
alpha = 0.05

# Step 2: Compute descriptive statistics manually
n = len(exam_scores)
sample_mean = np.mean(exam_scores)
sample_std = np.std(exam_scores, ddof=1)  # ddof=1 for sample standard deviation
sem = sample_std / np.sqrt(n)
df = n - 1

# Step 3: Perform One-Sample t-test via scipy.stats
t_statistic, p_value = stats.ttest_1samp(exam_scores, population_mean)
critical_val = stats.t.ppf(1 - alpha / 2, df)  # Two-tailed critical t

print(f"Sample Size (n): {n}")
print(f"Sample Mean: {sample_mean:.2f}")
print(f"Hypothesized Mean (mu0): {population_mean:.2f}")
print(f"t-Statistic: {t_statistic:.4f}")
print(f"Degrees of Freedom (df): {df}")
print(f"Critical t-Value (alpha=0.05): {critical_val:.4f}")
print(f"p-Value: {p_value:.4f}")

# Step 4: Statistical Decision
if p_value < alpha:
    print(f"Decision: Reject H0. The mean score ({sample_mean:.2f}) differs significantly from {population_mean}.")
else:
    print(f"Decision: Fail to reject H0. The mean score ({sample_mean:.2f}) is not significantly different from {population_mean}.")

# Step 5: Visualization with Error Bar and Hypothesized Mean Line
plt.figure(figsize=(6, 4))
plt.bar(1, sample_mean, color='skyblue', width=0.4, label='Sample Mean', edgecolor='black')
plt.errorbar(1, sample_mean, yerr=sample_std, fmt='k', capsize=5, linewidth=1.5, label='Sample Std Dev')
plt.axhline(y=population_mean, color='r', linestyle='--', linewidth=1.8, label=f'Hypothesized Mean ({population_mean})')
plt.xlim(0.5, 1.5)
plt.xticks([1], ['Class Exam Scores'])
plt.ylabel('Score')
plt.title('One-Sample t-Test: Sample Mean vs Hypothesized Mean')
plt.legend()
plt.show()

# 3.2 Student Work Task from T_test.ipynb
print("\n--- Student Work Task Evaluation ---")
student_data = np.array([5.2, 5.4, 5.1, 5.6, 5.3, 5.8])
hypo_mean = 5.5

t_stat_sw, p_val_sw = stats.ttest_1samp(student_data, hypo_mean)
print(f"Student Data Mean: {np.mean(student_data):.2f}")
print(f"t-Statistic: {t_stat_sw:.4f}, p-Value: {p_val_sw:.4f}")
if p_val_sw < alpha:
    print("Decision: Reject H0.")
else:
    print("Decision: Fail to reject H0. Mean is not significantly different from 5.5.")
```

# Module 4: Two-Sample Independent $t$-Test & Welch's Variance Formulation
**Source Resources:** `Two_Sample_t_Test.pptx.pdf`, `SIM_Day_6 (1).ipynb`

### When to Use:
Use a **Two-Sample Independent $t$-Test** when comparing the means of **two distinct, independent groups** to evaluate if their true population means differ.
* **Equal Variances Assumed (Student's Pooled $t$-test):** Use `equal_var=True` when sample variances are homogeneous ($\sigma_1^2 \approx \sigma_2^2$).
* **Unequal Variances Assumed (Welch’s $t$-test):** Use `equal_var=False` when group variances differ. Welch's test is more robust and prevents inflated Type I error.
* **Confidence Interval (CI) Rule:**
  * If the 95% Confidence Interval for the difference $(\mu_1 - \mu_2)$ **contains 0**, fail to reject $H_0$ (no significant difference).
  * If the 95% CI **does not contain 0**, reject $H_0$ (statistically significant difference).

### Mathematical Formulation:
* **Pooled Variance ($s_p^2$):**
  $$s_p^2 = \frac{(n_1 - 1)s_1^2 + (n_2 - 1)s_2^2}{n_1 + n_2 - 2}, \quad df = n_1 + n_2 - 2$$
* **Standard Error (Pooled):**
  $$\text{SE}_{\text{pooled}} = \sqrt{s_p^2 \left(\frac{1}{n_1} + \frac{1}{n_2}\right)}$$
* **Welch's Standard Error:**
  $$\text{SE}_{\text{Welch}} = \sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}$$
* **Confidence Interval for Difference of Means:**
  $$\text{CI} = (\bar{x}_1 - \bar{x}_2) \pm t_{\text{crit}} \times \text{SE}$$


```python
# 4.1 Class Lab Implementation (SIM_Day_6 (1).ipynb)
# Step 1: Define two independent datasets
data1 = np.array([23.1, 24.3, 22.8, 23.7, 24.0, 23.5])  # Sample 1 (e.g., Treatment Group)
data2 = np.array([21.9, 22.5, 22.0, 21.8, 22.2, 21.7])  # Sample 2 (e.g., Control Group)

# Step 2: Sample sizes, sample means, and sample variances
n1, n2 = len(data1), len(data2)
m1, m2 = np.mean(data1), np.mean(data2)
mean_diff = m1 - m2

# Note: ddof=1 is required for sample variance (dividing by n - 1)
s1_sq, s2_sq = np.var(data1, ddof=1), np.var(data2, ddof=1)

# Step 3: Pooled variance calculation (SIM_Day_6 formula)
sp_sq = ((n1 - 1) * s1_sq + (n2 - 1) * s2_sq) / (n1 + n2 - 2)
std_err_pooled = np.sqrt(sp_sq * (1/n1 + 1/n2))
df_pooled = n1 + n2 - 2

# 95% Confidence interval using Student's t distribution
ci_pooled = stats.t.interval(0.95, df=df_pooled, loc=mean_diff, scale=std_err_pooled)

# Step 4: Perform Two-Sample Pooled t-Test (equal_var=True)
t_stat_pooled, p_val_pooled = stats.ttest_ind(data1, data2, equal_var=True)

# Step 5: Perform Welch's t-Test (equal_var=False) for comparison
t_stat_welch, p_val_welch = stats.ttest_ind(data1, data2, equal_var=False)

print(f"Sample 1 Mean: {m1:.2f}, Variance: {s1_sq:.4f}")
print(f"Sample 2 Mean: {m2:.2f}, Variance: {s2_sq:.4f}")
print(f"Mean Difference (Sample 1 - Sample 2): {mean_diff:.2f}")
print(f"Pooled Variance (sp^2): {sp_sq:.4f}")
print(f"Pooled Standard Error: {std_err_pooled:.4f}")
print(f"Degrees of Freedom (Pooled): {df_pooled}")
print(f"95% Confidence Interval of Difference: [{ci_pooled[0]:.2f}, {ci_pooled[1]:.2f}]")
print(f"Pooled t-Statistic: {t_stat_pooled:.4f}, p-Value: {p_val_pooled:.4e}")
print(f"Welch t-Statistic: {t_stat_welch:.4f}, p-Value: {p_val_welch:.4e}")

# Step 6: Statistical Decision based on p-value and CI
alpha = 0.05
if p_val_pooled < alpha:
    print("Decision: Reject H0. There is a statistically significant difference between the two groups.")
else:
    print("Decision: Fail to reject H0. No statistically significant difference.")

if ci_pooled[0] > 0 or ci_pooled[1] < 0:
    print("CI Check: 0 is NOT inside the 95% CI -> Confirms significant difference.")
else:
    print("CI Check: 0 IS inside the 95% CI -> Confirms no significant difference.")

# Step 7: Visual Comparison
plt.figure(figsize=(7, 4))
plt.bar(['Sample 1', 'Sample 2'], [m1, m2], yerr=[np.sqrt(s1_sq), np.sqrt(s2_sq)], 
        capsize=5, color=['cornflowerblue', 'lightcoral'], edgecolor='black', width=0.5)
plt.ylabel('Measured Value')
plt.title('Two-Sample t-Test: Comparing Means with Sample Std Dev')
plt.show()
```

# Module 5: Chi-Square Goodness-of-Fit & Independence Test
**Source Resources:** `Chi Square.pptx`, `chi_sqr_test.ipynb`

### When to Use:
1. **Goodness-of-Fit Test:** When you have a single categorical variable with $k$ categories and wish to verify if the observed sample frequencies ($O_i$) conform to a theoretical distribution (e.g., checking if a 6-sided die is fair, checking if coin tosses are 50/50, checking uniform consumer preferences).
2. **Test of Independence:** When evaluating a 2D contingency table to test whether two categorical variables are independent or significantly associated.

### Mathematical Formulation:
$$\chi^2 = \sum_{i=1}^{k} \frac{(O_i - E_i)^2}{E_i}$$
* Degrees of Freedom: $df = k - 1$ (for Goodness-of-Fit with $k$ categories and no estimated parameters).
* Under a uniform distribution null hypothesis: $E_i = \frac{\sum O_i}{k}$.
* Decision Rules:
  * Reject $H_0$ if $\chi^2_{\text{calculated}} > \chi^2_{\text{critical}}$ where $\chi^2_{\text{critical}} = \text{stats.chi2.ppf}(1 - \alpha, df)$.
  * Reject $H_0$ if $p\text{-value} < \alpha$ ($0.05$).


```python
# 5.1 Goodness-of-Fit: Class Example (chi_sqr_test.ipynb)
# Step 1: Observed frequencies
observed = np.array([47, 49, 40])  # E.g., preferences across 3 categories

# Step 2: Expected frequencies assuming uniform distribution
total = np.sum(observed)
k = len(observed)
expected = np.full_like(observed, total / k, dtype=float)
print("Observed Frequencies:", observed)
print(f"Total Count: {total}, Expected Uniform Count per Category: {expected[0]:.2f}")

# Step 3: Manual Chi-square statistic calculation
chi_square_stat = np.sum(((observed - expected) ** 2) / expected)
df = k - 1

# Step 4: Critical value and p-value calculation
alpha = 0.05
critical_value = stats.chi2.ppf(1 - alpha, df)
p_value_chi = 1 - stats.chi2.cdf(chi_square_stat, df)

# Using scipy.stats.chisquare for verification
scipy_stat, scipy_p = stats.chisquare(observed, expected)

print(f"Calculated Chi-Square Statistic: {chi_square_stat:.4f} (SciPy: {scipy_stat:.4f})")
print(f"Degrees of Freedom (df): {df}")
print(f"Critical Value (alpha=0.05): {critical_value:.4f}")
print(f"p-Value: {p_value_chi:.4f} (SciPy: {scipy_p:.4f})")

# Step 5: Conclusion
if chi_square_stat > critical_value:
    print("Decision: Reject H0. Observed data does NOT match expected uniform distribution.")
else:
    print("Decision: Fail to reject H0. Observed data matches the expected uniform distribution.")

# Step 6: Plotting Observed vs Expected Side-by-Side (Class Template Trick)
categories = range(1, len(observed) + 1)
plt.figure(figsize=(7, 4))
plt.bar(categories, observed, width=0.35, label='Observed (Oi)', color='steelblue', alpha=0.8)
# np.arange is required because Python range does not support adding a float offset directly
plt.bar(np.arange(1, len(observed) + 1) + 0.35, expected, width=0.35, label='Expected (Ei)', color='orange', alpha=0.8)
plt.xlabel('Category')
plt.ylabel('Frequency')
plt.title('Chi-Square Test: Observed vs Expected Frequencies')
plt.xticks(np.arange(1, len(observed) + 1) + 0.35 / 2, [f'Cat {i}' for i in categories])
plt.legend()
plt.show()

# 5.2 Lab Assignment 1 Problem: observed = [45, 35, 25]
print("\n--- Lab Assignment 1 Task Evaluation ---")
obs_assign = np.array([45, 35, 25])
exp_assign = np.full_like(obs_assign, np.sum(obs_assign) / len(obs_assign), dtype=float)
stat_assign, p_assign = stats.chisquare(obs_assign, exp_assign)
crit_assign = stats.chi2.ppf(1 - alpha, len(obs_assign) - 1)
print(f"Assignment 1 Chi2 Stat: {stat_assign:.4f}, Critical Value: {crit_assign:.4f}, p-Value: {p_assign:.4f}")
if stat_assign > crit_assign:
    print("Decision: Reject H0. Significant departure from uniform distribution.")
else:
    print("Decision: Fail to reject H0. Data fits uniform distribution.")
```

# Module 6: Monte Carlo Simulation & Stochastic Modeling
**Source Resources:** `Monte_Carlo_Simulation_Basics.pptx.pdf`, `Day_4_Monte_Carlo_Simulation.ipynb`

### When to Use:
Use **Monte Carlo Simulation** when:
1. An analytical or closed-form mathematical solution is impossible, intractable, or extremely complex.
2. The problem involves uncertainty, probability distributions, risk analysis, or multi-dimensional integration.
3. You need to estimate properties of a system by generating repeated pseudorandom samples and computing empirical averages (guided by the **Law of Large Numbers**).

### Mathematical Formulation ($\pi$ Estimation):
* A circle of radius $r = 1$ centered at $(0, 0)$ has area $\pi r^2 = \pi$.
* An inscribed quarter circle in the unit square $[0, 1] \times [0, 1]$ has area $\frac{\pi}{4}$.
* The unit square has area $1 \times 1 = 1$.
* If $N$ random coordinate pairs $(x, y)$ are drawn uniformly where $x, y \sim U(0, 1)$:
  $$\text{Condition inside quarter circle}: \quad x^2 + y^2 \le 1$$
  $$\frac{\text{Points Inside}}{\text{Total Points}} \approx \frac{\text{Area of Quarter Circle}}{\text{Area of Unit Square}} = \frac{\pi / 4}{1} = \frac{\pi}{4}$$
  $$\pi \approx 4 \times \frac{\text{Inside Points}}{N}$$
* As $N \to \infty$, the simulation error decreases proportionally to $\frac{1}{\sqrt{N}}$.


```python
# 6.1 Monte Carlo Pi Estimation Function (Day_4_Monte_Carlo_Simulation.ipynb)
def estimate_pi(num_points, plot=False):
    inside = 0
    x_inside, y_inside = [], []
    x_outside, y_outside = [], []
    
    for _ in range(num_points):
        x = random.random()  # Continuous uniform float in [0.0, 1.0)
        y = random.random()
        if x*x + y*y <= 1.0:  # Distance from origin <= radius 1
            inside += 1
            if plot and len(x_inside) < 2000:  # Sample for visualization
                x_inside.append(x)
                y_inside.append(y)
        else:
            if plot and len(x_outside) < 2000:
                x_outside.append(x)
                y_outside.append(y)
                
    pi_estimate = 4.0 * inside / num_points
    error = abs(pi_estimate - np.pi)
    print(f"Num Points: {num_points:6d} | Estimated Pi: {pi_estimate:.5f} | Error: {error:.5f}")
    
    if plot:
        plt.figure(figsize=(5, 5))
        plt.scatter(x_inside, y_inside, color='dodgerblue', s=2, label='Inside Circle (x^2+y^2 <= 1)')
        plt.scatter(x_outside, y_outside, color='coral', s=2, label='Outside Circle')
        # Plot quarter circle boundary
        theta = np.linspace(0, np.pi/2, 100)
        plt.plot(np.cos(theta), np.sin(theta), color='darkblue', linewidth=2, label='Circle Boundary')
        plt.xlim(0, 1)
        plt.ylim(0, 1)
        plt.title(f'Monte Carlo Pi Estimation (N={num_points})')
        plt.xlabel('x')
        plt.ylabel('y')
        plt.legend(loc='lower left')
        plt.show()
        
    return pi_estimate

# 6.2 Demonstrating Convergence via the Law of Large Numbers
print("--- Monte Carlo Pi Convergence Table ---")
for points in [1000, 5000, 10000, 50000, 100000]:
    estimate_pi(points)

# Run with visual scatter plot
estimate_pi(10000, plot=True)
```
