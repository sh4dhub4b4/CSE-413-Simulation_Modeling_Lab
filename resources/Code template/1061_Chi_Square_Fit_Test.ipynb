# Chi Square Fit Test
Chi-Square Goodness-of-Fit Test in Python

## Import Necessary Libraries

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

## Step 1: Observed data (frequencies in each category)

```python
d=61
```

```python
observed = np.array([20+d, 25+d%10, 30+d%7,35+d%5])  # Example: Replace with your data
```

null_H = Students have an equal preference for all  languages.

alt_H = Students don't have an equal preference for all  languages.

## Step 2: Expected frequencies (assuming uniform distribution)

```python
total = np.sum(observed)
expected = np.full_like(observed, total / len(observed))
print(total / len(observed))  # Uniform distribution
print('Expected',expected)
```

## Step 3: Chi-square statistic calculation

```python
chi_square_stat = np.sum(((observed - expected) ** 2) / expected)
```

## Step 4: Degrees of freedom

```python
df = len(observed) - 1
```

## Step 5: Calculate the p-value and critical value

```python
#p_value = 1 - stats.chi2.cdf(chi_square_stat, df)  # p-value
alpha = 0.05  # Significance level
critical_value = stats.chi2.ppf(1 - alpha, df)  # Critical value
```

## Display the results

```python
print(f'Chi-square statistic: {chi_square_stat:.2f}')
print(f'Degrees of freedom: {df}')
print(f'Critical value (0.05 significance level): {critical_value:.2f}')
#print(f'P-value: {p_value:.4f}')
```

## Conclusion

```python
if chi_square_stat > critical_value:
    print('Reject the null hypothesis. Observed data does not match expected distribution. So Students do not have an equal preference for all  languages.')
else:
    print('Fail to reject the null hypothesis. Observed data matches the expected distribution. So Students have an equal preference for all  languages.')
```
