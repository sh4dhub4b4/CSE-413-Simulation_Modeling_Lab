```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Task 1
# id = 1061
d1,d2=1,6
A=[
    [d1+2,d2+1],
    [2*d1,d2+2]
]
Matrix(A)
```

```python
# Task 2
# matA=Matrix(A)
# shapeA=matA.shape
# detA=matA.det()
# rankA=matA.rank()
# eigenvalsA=matA.eigenvals()
# invA=matA.inv() if detA!=0 else 'None'

matA = np.array(A)
shapeA = matA.shape
detA = np.linalg.det(matA)
rankA = np.linalg.matrix_rank(matA)
eigenvalsA = np.linalg.eigvals(matA)
invA = np.linalg.inv(matA) if detA!=0 else 'None'

print("Shape:", shapeA)
print("Determinant:", detA)
print("Rank:", rankA)
print("Eigenvalues:", eigenvalsA)
print("Inverse:\n")
display(Matrix(invA))
```

```python
# Task 3

B = [row[:] for row in A]
B[0][1]+=1 #changing the (0,1) element

print("A =")
display(Matrix(A))

print("B =")
display(Matrix(B))
```

```python
# Task 4
matB=np.array(B)
shapeB=matB.shape
detB=np.linalg.det(matB)
rankB=np.linalg.matrix_rank(matB)
eigenvalsB=np.linalg.eigvals(matB)
invB=np.linalg.inv(matB) if detB!=0 else 'None'

print("Shape:", shapeB)
print("Determinant:", detB)
print("Rank:", rankB)
print("Eigenvalues:", eigenvalsB)
print("Inverse:\n")
display(Matrix(invB))
```

### 1. How did the determinant change and why?

The determinant changed from **10** to **8** after increasing one value in the matrix. This happened because the determinant depends on all the elements of the matrix.

### 2. Did the rank change? Explain.

No, the rank did not change. Both matrices have a rank of **2** because their determinants are not zero.

### 3. How did the eigenvalues respond to the value change?

The eigenvalues changed slightly.

- **Matrix A:** = **10**, **1**

### 4. Is B easier or harder to invert than A? Why?

Both matrices are easy to invert because both have non-zero determinants.

However, **B** has a smaller determinant than **A**, so it is slightly closer to being singular and can be a little more sensitive to numerical errors.
